```python
"""Phase 1 -- reference curves.  Run, read the checks, look at the plot,
and only then run scan_phase2.py.  Cheap enough to re-run interactively."""
import warnings; warnings.filterwarnings('ignore')
import numpy as np
from scan import scan_reference, check_reference, plot_reference

CASES = {
    'LiH': dict(geometry='Li 0 0 0; H 0 0 {}', spin=0, charge=0,
                R=np.round(np.arange(0.8, 4.01, 0.05), 3)),
    # 'N2':  dict(geometry='N 0 0 0; N 0 0 {}', spin=0, charge=0,
    #             R=np.round(np.arange(0.8, 2.81, 0.02), 3)),
}
BASIS = 'sto-3g'

for name, c in CASES.items():
    print(f'\n===== {name} : {len(c["R"])} geometries =====', flush=True)
    recs = scan_reference(name, c['geometry'], c['R'], spin=c['spin'],
                          charge=c['charge'], basis_set=BASIS,
                          outfile=f'{name.lower()}_phase1.pkl')
    problems = check_reference(recs)
    print('\nCHECKS:', flush=True)
    for p in (problems or ['all clear']):
        print('  -', p)
    plot_reference(recs, title=f'{name} / {BASIS.upper()}',
                   savefig=f'{name.lower()}_phase1.png')
```

```python
"""Phase 2 -- subspace curves, reusing the references approved in Phase 1.
Long and unattended: run under nohup, writes incrementally."""
import warnings; warnings.filterwarnings('ignore')
import os, pickle, sys
import pandas as pd
from scan import scan_subspaces

# NAMES = sys.argv[1:] or ['LiH', 'N2']
NAMES =  ['LiH', 'N2']
for name in NAMES:
    pkl, csv = f'{name.lower()}_phase1.pkl', f'{name.lower()}_scan.csv'
    if not os.path.exists(pkl):
        print(f'{name}: {pkl} missing -- run scan_phase1.py first'); continue
    recs = pickle.load(open(pkl, 'rb'))

    done = set()
    if os.path.exists(csv):
        done = set(pd.read_csv(csv).Distance.unique())
        print(f'{name}: resuming, {len(done)} geometries already done')
    todo = [r for r in recs if r['distance'] not in done]

    for i, rec in enumerate(todo, 1):
        df = scan_subspaces([rec], verbose=True)
        df.to_csv(csv, mode='a', header=not os.path.exists(csv), index=False)
        print(f'{name} {rec["distance"]:.3f}  ({i}/{len(todo)})', flush=True)

    print(f'{name} -> {csv}  ({len(pd.read_csv(csv))} rows)', flush=True)
```
